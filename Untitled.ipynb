```python
import pandas as pd
```

```python
df=pd.read_csv(r'/home/admin1/Desktop/vivek/datasets/student_dropout_risk_dataset.csv')
```

```python
df
```

```python
print(df.to_string())
```

```python
X=df[['backlogs','attendence','privious_year_cgpa','internal_marks','assi','score_of_attentation']]
```

```python
y=df['risk_score_dropout']
```

```python
print(X)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python

print(y_test)
```

```python
print(len(y_test))
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
X_train_scaled=scaler.fit_transform(X_train)
```

```python
X_test_scaled=scaler.transform(X_test)
```

```python
print(X_test_scaled)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
model=RandomForestRegressor()
```

```python
model.fit(X_train_scaled,y_train)
```

```python
model.score(X_test_scaled,y_test)
```

```python
import joblib
```

```python
joblib.dump(model,'studentsriskmodelensemble.pkl')
```

```python
joblib.dump(scaler,'studentsriskscalerensemble.pkl')
```
